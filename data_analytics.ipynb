```python
# data_analytics.csv

# update the column names to match the names you have in your csv file for data_analytics

# create a pandas dataframe
dfa = pd.read_csv(
    '/content/data_analytics.csv',
    sep=',',
    names = ['Page title and screen class','Views','Active users','Event count','Bounce rate'],
    skiprows=1
)

print(dfa.head())
```

```python
# data_survey.csv

# update the column names to match the names you have in your csv file for data_survey

# create a pandas dataframe
dfs = pd.read_csv(
    '/content/data_survey.csv',
    sep=',',
    names = ['Time', 'KnowledgeOfMovies','Dune1Watch','Dune1Rating','Dune2Watch','Dune2Rating', 'WillYouWatch', '2026Movie'],
    skiprows=1
)

# Display the first few rows of the DataFrame
print(dfs.head())
```

```python
# how many rows and columns are in the dataframe dfa? the shape is (3,5)
# how many rows and columns are in the dataframe dfs? the shape is (12,8)
```

```python
# Analytics

# Do one or more explorations with dfa
# no interaction is required

# make a blank canvas

plt.stairs(dfa["Views"])
plt.title("Views by Page")
plt.xlabel("Page (Data Analytics Lab vs. Questionnaire vs. Analytics)")
plt.ylabel("Quantity of Views")
plt.show()
# etc.

# reveal everything
```

```python
# Survey

# Do one or more explorations with dfs
# no interaction is required

# make a blank canvas

dfs["2026Movie"].value_counts().plot(kind="bar")
plt.xticks(rotation=45)
plt.tight_layout()
plt.title("Views by Page")
plt.xlabel("Page (Data Analytics Lab vs. Questionnaire vs. Analytics)")
plt.ylabel("Quantity of Views")
plt.show()
# etc.


# reveal everything
```
